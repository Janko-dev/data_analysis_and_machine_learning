```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = lambda x: x**3
d_f = lambda x: 3*x**2

x = np.linspace(-15, 15, 500)

plt.plot(x, f(x))
plt.plot(x, d_f(x))
```

## Newton's method to find roots

```python
x0 = -10
path = [x0]

xt = x0
for t in range(10):
    xt = xt - f(xt) / d_f(xt)
    path.append(xt)

path = np.array(path)

plt.plot(x, f(x))
plt.scatter(path, f(path))
plt.plot(path, f(path))

path, f(path)
```

## Newton's method to find optima

```python
d2_f = lambda x: 6*x
```

```python
x0 = -10
path = [x0]

xt = x0
for t in range(10):
    xt = xt - d_f(xt) / d2_f(xt)
    path.append(xt)

path = np.array(path)

plt.plot(x, f(x))
plt.scatter(path, f(path))
plt.plot(path, f(path))

path, f(path)
```

## General gradient ascent/descent

```python
x0 = -10
path = [x0]

xt = x0
for t in range(10):
    xt = xt - 0.001 * d_f(xt)
    path.append(xt)

path = np.array(path)

plt.plot(x, f(x))
plt.scatter(path, f(path))
plt.plot(path, f(path))

path, f(path)
```
