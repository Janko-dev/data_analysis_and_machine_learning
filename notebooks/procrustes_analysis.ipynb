```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_s_curve

X, y = make_s_curve(n_samples=1000, noise=0.2)

X = X[:, [0, 2]]

plt.scatter(X[:, 0], X[:, 1], c=y, s=8)
```

```python
q, r = np.linalg.qr(np.random.randn(2, 2))

X_rot = X @ q

plt.scatter(X_rot[:, 0], X_rot[:, 1], c=y, s=8)
```

```python
U, s, Vt = np.linalg.svd(X_rot.T @ X)

U.shape, Vt.shape
```

```python
U @ Vt, q
```

```python
np.mean((U @ Vt - q)**2)
```

```python
X_rot_recon = X @ U @ Vt

plt.scatter(X_rot_recon[:, 0], X_rot_recon[:, 1], c=y, s=8)
```
