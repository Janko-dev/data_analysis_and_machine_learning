```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import Ridge, LinearRegression
```

```python
x = np.linspace(-3, 3, 100)
y = np.sin(x)/np.exp(x) + np.random.randn(x.shape[0])*.4

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3)
```

```python
A = np.polynomial.polynomial.polyvander(x, deg=50)

model = Ridge(alpha=1e-1, fit_intercept=False).fit(A, y)

y_pred = model.predict(A)

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3, label='Ground truth')
plt.plot(x, y_pred, label='Predictions')
plt.legend()
```

```python
A = np.polynomial.chebyshev.chebvander(x, deg=50)

model = Ridge(alpha=1e+1, fit_intercept=False).fit(A, y)

y_pred = model.predict(A)

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3, label='Ground truth')
plt.plot(x, y_pred, label='Predictions')
plt.legend()
```

```python
A = np.polynomial.legendre.legvander(x, deg=50)

model = Ridge(alpha=1e-1, fit_intercept=False).fit(A, y)

y_pred = model.predict(A)

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3, label='Ground truth')
plt.plot(x, y_pred, label='Predictions')
plt.legend()
```

```python
A = np.polynomial.hermite.hermvander(x, deg=50)

model = Ridge(alpha=1e-3, fit_intercept=False).fit(A, y)

y_pred = model.predict(A)

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3, label='Ground truth')
plt.plot(x, y_pred, label='Predictions')
plt.legend()
```

```python
A = np.random.binomial(x, deg=50)

model = Ridge(alpha=1e-3, fit_intercept=False).fit(A, y)

y_pred = model.predict(A)

plt.plot(x, y, alpha=.5)
plt.scatter(x, y, s=3, label='Ground truth')
plt.plot(x, y_pred, label='Predictions')
plt.legend()
```
